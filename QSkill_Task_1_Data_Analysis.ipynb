```python
# Install/import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries imported successfully!')
```

## 1. Load the CSV file


```python
# Upload the CSV in Google Colab first, then run this cell.
# If the CSV is in the same folder as this notebook, this line works directly.
df = pd.read_csv('student_performance.csv')

print('Dataset loaded successfully!')
df.head()
```

## 2. Understand the dataset


```python
print('Shape (rows, columns):', df.shape)
print('\nColumn names:')
print(df.columns.tolist())

print('\nDataset information:')
df.info()

print('\nStatistical summary:')
display(df.describe())
```

## 3. Check for missing values and duplicates


```python
print('Missing values in each column:')
print(df.isnull().sum())

print('\nNumber of duplicate rows:', df.duplicated().sum())
```

## 4. Calculate average marks and attendance


```python
avg_math = df['Math'].mean()
avg_science = df['Science'].mean()
avg_english = df['English'].mean()
avg_attendance = df['Attendance'].mean()

print(f'Average Math marks: {avg_math:.2f}')
print(f'Average Science marks: {avg_science:.2f}')
print(f'Average English marks: {avg_english:.2f}')
print(f'Average Attendance: {avg_attendance:.2f}%')
```

## 5. Create a total/average score for each student


```python
subjects = ['Math', 'Science', 'English']
df['Average_Marks'] = df[subjects].mean(axis=1)

display(df[['Name'] + subjects + ['Average_Marks', 'Attendance']].sort_values('Average_Marks', ascending=False).head(10))
```

## 6. Bar chart – Average marks by subject


```python
subject_averages = df[subjects].mean()

plt.figure(figsize=(8,5))
subject_averages.plot(kind='bar')
plt.title('Average Marks by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Marks')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 7. Bar chart – Top 10 students by average marks


```python
top10 = df.nlargest(10, 'Average_Marks').sort_values('Average_Marks')

plt.figure(figsize=(9,5))
plt.barh(top10['Name'], top10['Average_Marks'])
plt.title('Top 10 Students by Average Marks')
plt.xlabel('Average Marks')
plt.ylabel('Student')
plt.tight_layout()
plt.show()
```

## 8. Scatter plot – Attendance vs Average Marks


```python
plt.figure(figsize=(8,5))
plt.scatter(df['Attendance'], df['Average_Marks'])
plt.title('Attendance vs Average Marks')
plt.xlabel('Attendance (%)')
plt.ylabel('Average Marks')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 9. Correlation heatmap


```python
numeric_cols = ['Age', 'Math', 'Science', 'English', 'Attendance', 'Average_Marks']
correlation = df[numeric_cols].corr()

plt.figure(figsize=(9,6))
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

## 10. Find the highest and lowest average performers


```python
highest = df.loc[df['Average_Marks'].idxmax()]
lowest = df.loc[df['Average_Marks'].idxmin()]

print('Highest average performer:')
print(highest[['Name', 'Average_Marks', 'Attendance']])

print('\nLowest average performer:')
print(lowest[['Name', 'Average_Marks', 'Attendance']])
```

## 11. Data-driven observations


```python
attendance_corr = df['Attendance'].corr(df['Average_Marks'])
best_subject = subject_averages.idxmax()
best_subject_avg = subject_averages.max()
lowest_subject = subject_averages.idxmin()
lowest_subject_avg = subject_averages.min()

print(f'1. The average {best_subject} score is the highest among the three subjects at {best_subject_avg:.2f}.')
print(f'2. The average {lowest_subject} score is the lowest among the three subjects at {lowest_subject_avg:.2f}.')
print(f'3. The correlation between attendance and average marks is {attendance_corr:.2f}.')
print('4. The scatter plot helps visualize whether higher attendance is associated with higher average marks.')
print('5. The heatmap shows the strength and direction of relationships among the numerical variables.')
```

## 12. Conclusion

The dataset was successfully loaded and analyzed using Pandas. Basic statistics were calculated and visualizations were created using Matplotlib and Seaborn. The bar charts compare subject performance, the scatter plot examines the relationship between attendance and marks, and the heatmap summarizes correlations among numerical variables. These results provide a simple data-driven view of student performance.
